```python
from sandygrad.scalar import Scalar
from sandygrad.nn import *
```

```python
# Train on the XOR problem
X = [[0, 0], [1, 0], [0, 1], [1, 1]]
y = [-1, 1, 1, -1] # -1:false, 1:true
```

```python
model = MLP(2, [10, 1])
print(model)
```

```python
lr = 0.05

for epoch in range(100):
    loss = []

    preds = []
    for Xi, yi in zip(X, y):
        #model.reset_dead_count()
        pred = model(Xi)
        preds.append(pred.data)
        #print(f"pred {pred}, ", end="")
        loss += [(1 - pred * yi).ReLU()]
    
    #print(model.zeros)

    total_loss = sum(loss)

    model.zero_grad()   
    total_loss.backward()

    for p in model.params():
        #print(f"{p.grad},", end="")
        p.data -= lr * p.grad

    print(f"epoch {epoch}, loss {total_loss.data}, preds {preds}", end="")

    print()
```
